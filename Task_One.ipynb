```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("datasaurus.csv")
df.head()
```

```python
def calculate_group_stats(group):
    return pd.Series({
        'x_mean': group['x'].mean(),
        'x_std': group['x'].std(),
        'y_mean': group['y'].mean(),
        'y_std': group['y'].std(),
        'x_y_corr': group['x'].corr(group['y'])
    })
```

```python
result = df.groupby('dataset').apply(calculate_group_stats).reset_index()
```

```python
result
```

 <h4> Every group have the same measure of central tendency, std and correlation for columns x and y </h4>

```python
g = sns.FacetGrid(df, col='dataset', col_wrap=4, height=3, aspect=1.2)
g.map(sns.scatterplot, 'x', 'y')

plt.subplots_adjust(top=0.9)
g.fig.suptitle("Scatter Plots Across All 13 Groups")

plt.show()
```

<h4> Only measures of central tendency, dispersion, and correlation can be misleading because they hide the true shape and distribution of data. Therefore, scatter plots should always be used alongside numerical metrics to reveal underlying patterns </h4>
