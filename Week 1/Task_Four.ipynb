```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv"
df = pd.read_csv(url)
df.head()
```

```python
# Assuming `df` is your pandas DataFrame loaded with the Auto MPG dataset
plt.figure(figsize=(9, 5))
sns.lineplot(
    data=df,
    x="model_year",
    y="mpg",
    marker="o",
    linewidth=2.5,
    color="#2b5c8f",
    errorbar=("ci", 95),
)

plt.title(
    "Average Fuel Efficiency (MPG) Trend (1970–1982)",
    fontsize=13,
    fontweight="bold",
    pad=12,
)
plt.xlabel("Model Year", fontsize=11)
plt.ylabel("Miles Per Gallon (MPG)", fontsize=11)
plt.xticks(
    range(df["model_year"].min(), df["model_year"].max() + 1),
    labels=[f"'{y}" for y in range(70, 83)],
)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python


# Map numeric origin codes if dataset uses integers (1: USA, 2: Europe, 3: Japan)
origin_map = {1: "usa", 2: "europe", 3: "japan"}
df["origin_label"] = df["origin"].map(origin_map).fillna(df["origin"])

plt.figure(figsize=(8, 5))

# Create boxplot to display distribution metrics (median, IQR, outliers)
sns.boxplot(
    data=df,
    x="origin_label",
    y="mpg",
    palette="Set2",
    width=0.4,
    boxprops=dict(alpha=0.85),
)

# Overlay individual data points for granular visibility
sns.stripplot(
    data=df,
    x="origin_label",
    y="mpg",
    color="black",
    alpha=0.3,
    jitter=0.2,
    size=5,
)

plt.title(
    "Fuel Efficiency (MPG) Distribution by Region of Origin",
    fontsize=13,
    fontweight="bold",
    pad=12,
)
plt.xlabel("Region of Origin", fontsize=11)
plt.ylabel("Miles Per Gallon (MPG)", fontsize=11)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python


# Assuming `df` is your pandas DataFrame containing the Auto MPG dataset
plt.figure(figsize=(8, 5))

# Plot bivariate relationship between continuous variables with fitted linear trendline
sns.regplot(
    data=df,
    x="weight",
    y="mpg",
    scatter_kws={"alpha": 0.6, "color": "#2b5c8f", "s": 30},
    line_kws={"color": "#e41a1c", "linewidth": 2},
)

plt.title(
    "Relationship Between Vehicle Weight and Fuel Efficiency (MPG)",
    fontsize=13,
    fontweight="bold",
    pad=12,
)
plt.xlabel("Vehicle Weight (lbs)", fontsize=11)
plt.ylabel("Miles Per Gallon (MPG)", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

<h4> The third chart shows the vehicle weight is being reduced over time in order to achieve more fuel efficiency</h4>

```python
# Assuming `df` is your pandas DataFrame loaded with the Auto MPG dataset
plt.figure(figsize=(8, 5))

# Count the frequency of each cylinder configuration
cyl_counts = df["cylinders"].value_counts().sort_index()

ax = sns.barplot(
    x=cyl_counts.index, y=cyl_counts.values, palette="Blues_r"
)

# Annotate exact count values on top of each bar
for p in ax.patches:
    ax.annotate(
        f"{int(p.get_height())}",
        (p.get_x() + p.get_width() / 2.0, p.get_height()),
        ha="center",
        va="center",
        xytext=(0, 5),
        textcoords="offset points",
        fontsize=10,
        fontweight="bold",
    )

plt.title(
    "Distribution of Engine Cylinders Across the Vehicle Fleet",
    fontsize=13,
    fontweight="bold",
    pad=12,
)
plt.xlabel("Number of Cylinders", fontsize=11)
plt.ylabel("Vehicle Count", fontsize=11)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python

# Assuming `df` is your pandas DataFrame loaded with the Auto MPG dataset
plt.figure(figsize=(9, 5))

# Plot the full probability density of car weights across each model year
sns.violinplot(
    data=df,
    x="model_year",
    y="weight",
    palette="Blues",
    inner="quartile",
    cut=0,
)

plt.title(
    "Vehicle Weight Distribution Shift Across Model Years (1970–1982)",
    fontsize=13,
    fontweight="bold",
    pad=12,
)
plt.xlabel("Model Year", fontsize=11)
plt.ylabel("Vehicle Weight (lbs)", fontsize=11)
plt.xticks(
    ticks=range(13),
    labels=[f"'{y}" for y in range(70, 83)],
)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

<h4> I have pick box plot chart as my weakest one and use the voilen plot inplace which shows the distribution and clusters or groups within data along with the statistics like , median, 25 percentile and 75 percentile </h4>

```python
origin_map = {1: "USA", 2: "Europe", 3: "Japan"}
df["origin_label"] = (
    df["origin"].map(origin_map).fillna(df["origin"]).str.title()
)

plt.figure(figsize=(9, 5.5))
palette = {"Usa": "#2b5c8f", "Europe": "#2ca02c", "Japan": "#d62728"}

# Horizontal Violin Plot showing full density profile and quartile lines
sns.violinplot(
    data=df,
    y="origin_label",
    x="mpg",
    palette=palette,
    inner="quartile",
    cut=0,
    density_norm="width",
)

# Non-obstructive stripplot overlay for individual sample transparency
sns.stripplot(
    data=df,
    y="origin_label",
    x="mpg",
    color="black",
    alpha=0.35,
    jitter=0.2,
    size=4,
)

plt.title(
    "Fuel Efficiency (MPG) Distribution by Region of Origin",
    fontsize=13,
    fontweight="bold",
    pad=12,
)
plt.xlabel("Miles Per Gallon (MPG)", fontsize=11)
plt.ylabel("Region of Origin", fontsize=11)
plt.grid(axis="x", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
